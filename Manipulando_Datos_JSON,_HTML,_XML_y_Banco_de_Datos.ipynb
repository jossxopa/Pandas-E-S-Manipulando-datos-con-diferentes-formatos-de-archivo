### Datos anidados en JSON

Cuando existen en los datos de JSON más datos dentro de un parametro en forma de lista se debe normalizar para que cada dato tome el lugar en una columna.

```python
datos_pacientes2 = pd.read_json("pacientes_2.json")
```

```python
datos_pacientes2 #Podemos observar que en pacientes se engloba toda la información de un paciente en una sola columna, pero queremos cada dato en una columna diferente
```

```python

datos_normalizados = pd.json_normalize(datos_pacientes2["Pacientes"])
datos_normalizados

# El resultado será la separación de la información
```

## Exportar el archivo

Podemos exportarlo y guardalo usando el método to_json, pero solo lo podríamos leer en ese formato si quisieramos enviarlo en otro formato de archivo lo hacemos con to_csv o to_xmls

```python
datos_normalizados.to_json("Historio_de_Pacientes.json", index=False)
```

# Banco de datos

```python
import sqlalchemy

from sqlalchemy import create_engine, MetaData, Table, inspect, text
```

Importar los métodos necesarios de la biblioteca SQLAlchemy:

* createEngine: Para crear el motor de la base de datos.
* MetaData: Para guardar la información de las tablas (nombres, atributos, tipos de datos).
* Table: Para trabajar y consultar las tablas.
* Inspect: Para inspeccionar bancos de datos y tablas.
* text: Para convertir las consultas en un tipo de dato de texto que se pueda entender.
* **createEngine('sqlite:///:memory:')**: para crear un motor SQLite que se ejecute en la memoria local de Google Cloud.
Guardar el resultado en una variable llamada engine para referenciarlo posteriormente.

```python
engine = create_engine("sqlite:///:memory:")

# El objeto 'engine' es ahora el punto de acceso a la base de datos.
# Se puede usar para crear tablas, insertar datos y realizar consultas.
```

```python
archivo = "clientes_banco.csv"
datos = pd.read_csv(archivo)
datos
```

```python
# Vamos a exportar el DataFrame a una tabla SQL

datos.to_sql("clientes", engine, index=False)

# Crear un inspector para verificar la creación de una tabla

inspector = inspect(engine)

# Imprimir los nombres de las tablas en la base de datos

print(inspector.get_table_names())
```

## Crear una tabla SQL con datos filtrados

En este caso solo queremos a los empleados, así que de la columna "Categoria_de_renta" escogeremos solo a los empleados

```python
query = "SELECT * FROM clientes WHERE Categoria_de_renta = 'Empleado'"

"""
Aquí dice: Selecciona todas* la columnas de la columna clientes donde vas a elegir Empleado de la
columna Categoria_de_renta
"""

empleados = pd.read_sql(sql=text(query), con = engine.connect()) #el text es delmétodo que hicimos al principio cuando importamos sqlalchemy

#Con significa: conexión
# Dice la línea: Conectate al motor al motor sqllite (engine) y en ese motor ejecuta la query que hicimos donde dice que
# se exraiga la información de empleado, es decir aplicar el filtro que hicimos
```

```python
empleados
```

Ya tenemos nuestro DataFrame de empleados ahora lo pasaremos a una tabla sql

```python
empleados.to_sql("empleados", con = engine.connect(), index=False)
```

## ¿Cómo leer la nueva tabla?

A diferencia de los otro ejemplos, la nueva tabla se lee diferente, ya que si usamos read, estaríamos leyendo una query(el filtro), pero si queremos leer la tabla completa debemos usar el método "table"

```python
pd.read_sql_table("empleados", con = engine.connect())
```

## Ver la tabla pero eligiendo ciertas columnas

Para ello solo agregamos el método columns, corchetes [] y, en medio de estos las columnas que queramos mostrar

```python
pd.read_sql_table("empleados", con = engine.connect(), columns=["ID_Cliente", "Grado_estudio", "Edad", "Rendimiento_anual"])
```

## Modificar los datos de la tabla

### Borrar *datos*

```python
query = "SELECT * FROM clientes"

pd.read_sql(sql=text(query), con = engine.connect())
```

```python
# Vamos a eliminar el eliminar el primer ID query: 5008804
# Primero importamos SQLAlchemy error para que nos informe si hay un error al querer manipular

from sqlalchemy.exc import SQLAlchemyError
```

```python
query = "DELETE FROM clientes WHERE ID_Cliente = 5008804"

try:
  r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
  print(e)
else:
  print("Número de registros eliminados: ", r_set.rowcount)
```

```python
# Vamos a corroborar si se elimino

pd.read_sql_table("clientes", con = engine.connect())

#Podemos observar que ya no existe
```

### Actualizar datos

```python
query = "UPDATE clientes SET Grado_estudio = 'Nivel superior' WHERE ID_Cliente = 5008808"

# SET especifica que columna va a modificar
# Actuaiza de la tabla clientes en la columna grado estudios y se le va a asignar el valor Nivel superior al ID 5008808

try:
  r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
  print(e)
else:
  print("Número de registros modificados: ", r_set.rowcount)
```

```python
# Verificamos el cambio

pd.read_sql_table("clientes", con = engine.connect())

#Vemos que si cambió
```

## Conocer cuantas tablas tenemos

El método **Inspector**

* El inspector de SQLAlchemy es una herramienta muy útil que te permite obtener información detallada sobre la base de datos y sus objetos.
* Además de obtener los nombres de las tablas, el inspector puede proporcionar información sobre:
* Columnas de una tabla: Nombres, tipos de datos, restricciones, etc.
* Índices: Nombres, columnas que lo componen, si es único o no.
* Restricciones: Claves primarias, claves foráneas, restricciones de unicidad etc.
* Vistas: Definición de la vista.
* Triggers: Definición del trigger.

el inspector es una herramienta muy versátil que te permite obtener una visión completa de la estructura de la base de datos.

```python
inspector = inspect(engine)
nombres_tablas = inspector.get_table_names()
print(nombres_tablas)
```
